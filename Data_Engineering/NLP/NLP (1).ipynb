```python
def remove_stopwords(text):
    tokens = word_tokenize(text)
    filtered_tokens = [word for word in tokens if word not in stop_words]
    return ' '.join(filtered_tokens)
```

```python
df['stopwords_removed'] = df['num_removed'].apply(remove_stopwords)
```

```python
def remove_url(text):
  return re.sub(r'http\S+|www.\S+','',text)
df['url_removed']=df['stopwords_removed'].apply(remove_url)
```

```python
def remove_hashtag(text):
  return re.sub(r'#\w+','',text)
df['hashtag_removed']=df['url_removed'].apply(remove_hashtag)
```

```python
def remove_emojis(text):
  emoticon_pattern=r"[:;=]-?[)(|DPOo0/\\]+"
  text=re.sub(emoticon_pattern,'',text)
  return text
df['emoji_removed']=df['hashtag_removed'].apply(remove_emojis)
```

```python
def remove_extra_whitespaces(text):
  return ' '.join(text.split())
df['extra_space_removed']=df['emoji_removed'].apply(remove_extra_whitespaces)
```

```python
from nltk.stem import PorterStemmer
ps=PorterStemmer()
```

```python
words=["playing","played","running","crying","caring"]
```

```python
def stemming_text(text):
  tokens=nltk.word_tokenize(text)
  stemmed_words=[ps.stem(word) for word in tokens]
  return ' '.join(stemmed_words)
#df["stemmed_words"]=df['extra_space_removed'].apply(stemming_text)
```

```python
df["stemmed_words"]=df['extra_space_removed'].apply(stemming_text)
```

```python
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet
nltk.download('wordnet')
lemma=WordNetLemmatizer()
```

```python
def lemmatize_text(text):
  tokens=nltk.word_tokenize(text)
  lemma_words=[lemma.lemmatize(word,wordnet.VERB) for word in tokens]
  return ' '.join(lemma_words)
```

```python
df['lemma_words']=df['extra_space_removed'].apply(lemmatize_text)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train,y_test=train_test_split(df['lemma_words'],df['Label'],test_size=0.1,random_state=12)
```

```python
cv=CountVectorizer(ngram_range=(1,2))
Xtrain_cv=cv.fit_transform(X_train)
Xtest_cv=cv.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression(max_iter=100)
lr.fit(Xtrain_cv,y_train)
y_pred = lr.predict(Xtest_cv)
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
```

```python
print("Accuracy_Score is:", accuracy_score(y_test,y_pred))
print("precision_score is:", precision_score(y_test,y_pred))
print("recall_score is:", recall_score(y_test,y_pred))
print("f1_score is:", f1_score(y_test,y_pred))
```
