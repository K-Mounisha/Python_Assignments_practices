### Problem 7.1: Collatz Conjecture (Medium)

The Collatz conjecture is one of the most famous unsolved problems in mathematics. The rules are:
1. Start with any positive integer n
2. If n is even, divide it by 2
3. If n is odd, multiply it by 3 and add 1
4. Repeat until you reach 1

Write a program that:
- Takes a number as input
- Shows the complete sequence until it reaches 1
- Counts how many steps it took

Example: Starting with 6: 6 → 3 → 10 → 5 → 16 → 8 → 4 → 2 → 1 (8 steps)

**Hint:** While n != 1: if n%2==0, n=n//2, else n=3*n+1. Count steps and print each value.

````python
# Your code here
n = int(input("Enter a positive number: "))

steps = 0

print("Collatz sequence:")

while n != 1:
    print(n, end=" → ")
    
    if n % 2 == 0:
        n = n // 2
    else:
        n = 3 * n + 1
    
    steps += 1

print(1)  
print("Total steps:", steps)
````

### Problem 7.2: Caesar Cipher (Hard)

Write a program that implements a Caesar cipher to encrypt and decrypt messages.

The Caesar cipher shifts each letter by a fixed number of positions in the alphabet.

Example: With shift=3
- 'A' becomes 'D'
- 'Hello' becomes 'Khoor'

Your program should:
1. Ask for a message and shift value
2. Provide options to encrypt or decrypt
3. Handle both uppercase and lowercase letters
4. Keep non-alphabetic characters unchanged

**Hint:** Loop through each character. Use ord() to get ASCII value, shift it, use chr() to convert back. Use modulo 26 for wrapping. Check if char.isalpha().

````python
# Your code here
message = input("Enter your message: ")
shift = int(input("Enter shift value: "))
choice = input("Type 'e' to Encrypt or 'd' to Decrypt: ")

result = ""
if choice == 'd':
    shift = -shift

for ch in message:
    if ch.isalpha():
        if ch.isupper():
            base = ord('A')
            result += chr((ord(ch) - base + shift) % 26 + base)
        else:
            base = ord('a')
            result += chr((ord(ch) - base + shift) % 26 + base)
    else:
        result += ch   

print("Result:", result)
````

````python
# Your code here
balance = 10000
pin = "1234"
attempts = 3
history = []
while attempts > 0:
    entered_pin = input("Enter your PIN: ")
    if entered_pin == pin:
        print("Login Successful!\n")
        break
    else:
        attempts -= 1
        print("Wrong PIN! Attempts left:", attempts)

if attempts == 0:
    print("Account Locked! Try later.")
else:
    while True:
        print("\n===== ATM MENU =====")
        print("1. Check Balance")
        print("2. Withdraw Money")
        print("3. Deposit Money")
        print("4. Change PIN")
        print("5. Exit")

        choice = input("Enter your choice: ")
        if choice == "1":
            print("Current Balance:", balance)
            history.append("Checked Balance")
        elif choice == "2":
            try:
                amount = float(input("Enter amount to withdraw: "))
                if amount <= 0:
                    print("Invalid amount!")
                elif amount > balance:
                    print("Insufficient balance!")
                else:
                    balance -= amount
                    print("Withdrawal Successful!")
                    history.append(f"Withdrawn {amount}")
            except:
                print("Invalid input!")
        elif choice == "3":
            try:
                amount = float(input("Enter amount to deposit: "))
                if amount <= 0:
                    print("Invalid amount!")
                else:
                    balance += amount
                    print("Deposit Successful!")
                    history.append(f"Deposited {amount}")
            except:
                print("Invalid input!")
        elif choice == "4":
            old_pin = input("Enter old PIN: ")
            if old_pin == pin:
                new_pin = input("Enter new PIN: ")
                if new_pin.isdigit() and len(new_pin) == 4:
                    pin = new_pin
                    print("PIN changed successfully!")
                    history.append("Changed PIN")
                else:
                    print("PIN must be 4 digits!")
            else:
                print("Incorrect old PIN!")
        elif choice == "5":
            print("Thank you for using ATM!")
            break

        else:
            print("Invalid choice!")
        if len(history) > 5:
            history.pop(0)
        print("\nLast Transactions:")
        for item in history:
            print("-", item)
````

### Problem 11.8: Sieve of Eratosthenes (Hard)

**Hint:** Create boolean list (all True). For each number i from 2 to √n, mark all multiples of i as False (not prime). Print numbers that remain True.

````python
# Your code here
n = int(input("Enter a number: "))
prime = [True] * (n + 1)
if n >= 0:
    prime[0] = False
if n >= 1:
    prime[1] = False

i = 2
while i * i <= n:
    if prime[i]:
        j = i * i
        while j <= n:
            prime[j] = False
            j += i
    i += 1
print("Prime numbers up to", n, ":")
for i in range(2, n + 1):
    if prime[i]:
        print(i, end=" ")
````

### Problem 11.10: Sudoku Validator (Hard)

Validate if a given 9×9 Sudoku solution is correct.

**Hint:** Check three conditions: (1) Each row has unique 1-9, (2) Each column has unique 1-9, (3) Each 3x3 box has unique 1-9. Use sets to check uniqueness.

````python
# Your code here
print("Enter Sudoku board (9 rows, 9 numbers each separated by space):")

board = []
for i in range(9):
    row = list(map(int, input().split()))
    board.append(row)

valid = True
for row in board:
    if set(row) != set(range(1, 10)):
        valid = False
        break
if valid:
    for col in range(9):
        column = []
        for row in range(9):
            column.append(board[row][col])
        if set(column) != set(range(1, 10)):
            valid = False
            break
if valid:
    for box_row in range(0, 9, 3):
        for box_col in range(0, 9, 3):
            box = []
            for i in range(3):
                for j in range(3):
                    box.append(board[box_row + i][box_col + j])
            if set(box) != set(range(1, 10)):
                valid = False
                break
if valid:
    print("Valid Sudoku Solution")
else:
    print("Invalid Sudoku Solution")
````
