```python
a=float(input("Enter the triangle side A: "))
b=float(input("Enter the triangle side B: "))
c=float(input("Enter the triangle side C: "))
if a+b>c and b+c>a and c+a>b:
  if a==b==c:
    print(("Triangle is a equilateral"))
  elif a==b or b==c or c==a:
    print("Triangle is a isosceles")
  else:
    print("Triangle is a scalene")
```

```python
weight=float(input("Enter your weight: "))
height=float(input("Enter your height: "))
BMI=weight/height**2
if BMI<18.5:
  print("Underweight")
elif BMI>18.5 and BMI<24.9:
  print("Normal weight")
elif BMI>25 and BMI<29.9:
  print("Over weight")
elif BMI>30:
  print("Obese")
else:
  print("BMI is high")
```

```python
#to find how many no of even numbers and odd numbers
even_count=0
odd_count=0
for i in range(1,11):
  if i%2==0:
    even_count+=1
  else:
    odd_count+=1
print(f"Even count is {even_count}")
print(f"Odd count is {odd_count}")
```

```python
#to count the sum of even numbers
number=(1,11,2,12,3,13,4,15,6,16,7,17,8,18,9,19,10)
even_count=0
odd_count=0
for num in number:
  if num%2==0:
    even_count+=num
  else:
    odd_count+=num
print(f"Even sum is {even_count}")
print(f"Odd sum is {odd_count}")
```

```python
#prime number
n=int(input("Enter a number: "))
if n<=1:
  print("NOt a prime number")
else:
  for i in range(2,n):
    if n%i==0:
      print("Not a prime number")
      break
  else:
    print("Prime number")
```

```python
#sum of digits
n=int(input("Enter a number: "))
total=0
while n>0:
  digit=n%10
  total+=digit
  n=n//10
print("Sum of digits: ", total)
```

```python
#palindrome checker
n=int(input("Enter a number: "))
original=n
reverse=0
while n>0:
  digit=n%10
  reverse=reverse*10+digit
  n=n//10
if original==reverse:
  print("Palindrome number")
else:
  print("Not a palindrome number")
```

```python
#pyramid
n=5
for i in range(1,n+1):
  for j in range(n-i):
    print(" ",end=" ")
  for k in range(2*i-1):
    print("*",end=" ")
  print()
```
