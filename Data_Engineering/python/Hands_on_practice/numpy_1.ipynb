```python
a=np.array([10,20,30,40])
b=np.array([1,2,3,4])
```

```python
print("Array A:", a)
print("Array B:", b)
```

```python
print("\nAdditon (np.add):")
print(np.add(a, b))
```

```python
print("\nSubtraction (np.subtract):")
print(np.subtract(a,b))
```

```python
print("\nMultiplication (np.multiply)")
print(np.multiply(a,b))
```

```python
print("\nDivision (np.divide):")
print(np.divide(a, b))
```

```python
print("\nPower (np.power):")
print(np.power(a,b))
```

```python
print("\nModulus (np.mod):")
print(np.mod(a,b))
```

```python
print("\nFloor Division (np.floor_divide):")
print(np.floor_divide(a,b))
```

```python
print("\nAbsolute  Difference(np.divide):")
print(np.abs(a-b))
```
