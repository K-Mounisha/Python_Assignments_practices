```python
import matplotlib.pyplot as plt
```

```python
import pandas as pd
```

```python
import numpy as np
```

```python
np.random.seed(42)
data={
        "Employee_id": range(1,101),
        "Age": np.random.randint(21,60,100),
        "Salary": np.random.randint(20000,120000,100),
        "Experience": np.random.randint(1,20,100),
        "Department": np.random.choice(["IT", "HR", "Finance", "Sales"], 100),
        "Performance_Score": np.random.randint(1,5,100)
}
```

```python
data
```

```python
df=pd.DataFrame(data)
print(df)
```

```python
print(df.head())
```

```python
print(df.tail())
```

```python
#lineplot
plt.figure(figsize=(20,5))
plt.plot(df["Employee_id"], df["Salary"], color='red')
```

```python
plt.xlabel("Employee_id")
plt.ylabel("Salary")
```

```python
plt.title("Salary trend")
```

```python
plt.show()
```

```python
plt.figure(figsize=(20,5))
plt.plot(df["Employee_id"], df["Salary"],color='purple')
plt.xlabel("Emplpoyee id")
plt.ylabel("Salary")
plt.title("Salary Trend")
plt.show()
```

```python
plt.figure(figsize=(20,10))
plt.plot(df["Employee_id"], df["Salary"],color='purple',marker='o')
plt.xlabel("Emplpoyee id")
plt.ylabel("Salary")
plt.title("Salary Trend")
plt.show()
```

```python
plt.figure(figsize=(20,10))
plt.plot(df["Employee_id"], df["Salary"],color='purple',marker='o',markersize=20)
plt.xlabel("Emplpoyee id")
plt.ylabel("Salary")
plt.title("Salary Trend")
plt.show()
```

```python
plt.figure(figsize=(20,10))
plt.subplot(2,2,1)
plt.plot(df["Employee_id"], df["Salary"])
plt.subplot(2,2,2)
plt.plot(df["Employee_id"], 1/df["Salary"],color='r')
plt.subplot(2,2,3)
plt.plot(df["Employee_id"], 1/df["Salary"],color='b')
plt.subplot(2,2,4)
plt.plot(df["Employee_id"], 1/df["Salary"],color='g')
```

```python
#Scatter plot
plt.figure()
plt.scatter(df["Age"], df["Salary"],color='g',alpha=0.5)
plt.title("Age vs Salary")
plt.show()
```

```python
#histogram
plt.figure()
plt.hist(df["Salary"])
plt.title("Salary Distribution")
plt.show()
```
