mode 'rb' and 'wb':used for images

```python
#reading a binary data
with open("image.jpg", 'rb') as f:
    data=f.read()
    print(type(data))
    print('Size: ',len(data), "bytes")
    print("First 4bytes:",data[:4])
```

```python
#copying binary file
with open("image.jpg", "rb") as src:
    with open("photo.jpg","wb") as dst:
        dst.write(src.read())
```

writelines()
f.writelines(list)-write each element of the list to the file one after another


```python
fruits=["Apple\n", "Banana\n", "Cherry\n", "Data\n"]
with open('fruits.txt', 'w') as f:
    f.writelines(fruits)
```

```python
fruits=["Apple", "Banana", "cherry", "Data"]
with open('fruits_2.txt','w') as f:
    f.writelines(fruits)
```
